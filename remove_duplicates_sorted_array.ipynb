```python

class Solution:
    def removeDuplicates(self, nums):
        if not nums:
            return 0

        k = 1

        for i in range(1, len(nums)):
            if nums[i] != nums[k - 1]:
                nums[k] = nums[i]
                k += 1

        return k


solution = Solution()

nums1 = [1, 1, 2]
k1 = solution.removeDuplicates(nums1)
print("Count:", k1, "Array:", nums1[:k1])

nums2 = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k2 = solution.removeDuplicates(nums2)
print("Count:", k2, "Array:", nums2[:k2])

nums3 = [1, 2, 3, 4]
k3 = solution.removeDuplicates(nums3)
print("Count:", k3, "Array:", nums3[:k3])

nums4 = [5, 5, 5, 5]
k4 = solution.removeDuplicates(nums4)
print("Count:", k4, "Array:", nums4[:k4])
```
