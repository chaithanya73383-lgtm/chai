```python


import heapq

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution:
    def mergeKLists(self, lists):
        heap = []

        for i, node in enumerate(lists):
            if node:
                heapq.heappush(heap, (node.val, i, node))

        dummy = ListNode()
        current = dummy

        while heap:
            val, i, node = heapq.heappop(heap)

            current.next = node
            current = current.next

            if node.next:
                heapq.heappush(
                    heap, (node.next.val, i, node.next)
                )

        return dummy.next
```

```python

def make_list(values):
    dummy = ListNode()
    current = dummy

    for value in values:
        current.next = ListNode(value)
        current = current.next

    return dummy.next


def print_list(node):
    result = []

    while node:
        result.append(node.val)
        node = node.next

    print(result)


lists = [
    make_list([1, 4, 5]),
    make_list([1, 3, 4]),
    make_list([2, 6])
]

answer = Solution().mergeKLists(lists)
print_list(answer)
```
