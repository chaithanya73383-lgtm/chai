```python

class Solution:
    def reverse(self, x):
        sign = -1 if x < 0 else 1
        reversed_num = int(str(abs(x))[::-1]) * sign

        if reversed_num < -(2**31) or reversed_num > 2**31 - 1:
            return 0

        return reversed_num


solution = Solution()

print(solution.reverse(123))
print(solution.reverse(-123))
print(solution.reverse(120))
print(solution.reverse(1534236469))
```
