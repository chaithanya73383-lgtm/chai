```python

class Solution:
    def isPalindrome(self, x):
        if x < 0:
            return False

        return str(x) == str(x)[::-1]


solution = Solution()

print(solution.isPalindrome(121))
print(solution.isPalindrome(-121))
print(solution.isPalindrome(10))
print(solution.isPalindrome(1221))
```
