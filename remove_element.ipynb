```python

class Solution:
    def removeElement(self, nums, val):
        k = 0

        for i in range(len(nums)):
            if nums[i] != val:
                nums[k] = nums[i]
                k += 1

        return k


solution = Solution()

nums1 = [3, 2, 2, 3]
k1 = solution.removeElement(nums1, 3)
print("Count:", k1, "Array:", nums1[:k1])

nums2 = [0, 1, 2, 2, 3, 0, 4, 2]
k2 = solution.removeElement(nums2, 2)
print("Count:", k2, "Array:", nums2[:k2])

nums3 = [1, 1, 1]
k3 = solution.removeElement(nums3, 1)
print("Count:", k3, "Array:", nums3[:k3])

nums4 = [1, 2, 3]
k4 = solution.removeElement(nums4, 4)
print("Count:", k4, "Array:", nums4[:k4])
```
