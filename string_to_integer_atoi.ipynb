```python


class Solution:
    def myAtoi(self, s):
        i = 0
        n = len(s)

        # Skip leading spaces
        while i < n and s[i] == " ":
            i += 1

        # Read the sign
        sign = 1

        if i < n and s[i] == "-":
            sign = -1
            i += 1
        elif i < n and s[i] == "+":
            i += 1

        # Read digits
        number = 0

        while i < n and s[i].isdigit():
            number = number * 10 + int(s[i])
            i += 1

        number *= sign

        # Apply 32-bit signed integer limits
        INT_MIN = -(2 ** 31)
        INT_MAX = (2 ** 31) - 1

        if number < INT_MIN:
            return INT_MIN

        if number > INT_MAX:
            return INT_MAX

        return number


solution = Solution()

print(solution.myAtoi("42"))
print(solution.myAtoi("   -42"))
print(solution.myAtoi("4193 with words"))
print(solution.myAtoi("words and 987"))
print(solution.myAtoi("-91283472332"))
print(solution.myAtoi("+123"))
print(solution.myAtoi(""))
print(solution.myAtoi("+-12"))
```
